# Сопоставление правдивых письменных историй и видео

Ноутбук читает результаты воспроизводимого пайплайна. Письменные истории подтверждены пользователем как правдивые; метки видеоответов не подменяются метками источников. Статистическая единица — респондент. Тексты и стенограммы остаются в локальных исключённых каталогах.

Публичная версия не содержит сохранённых результатов выполнения. Локальный запуск читает приватные производные данные; перед публикацией снова очистите выводы ячеек.


```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
out = root / 'outputs' / 'research'
assert out.exists(), 'Сначала запустите python scripts/analyze_research.py из корня проекта'
summary = json.loads((out / 'summary.json').read_text(encoding='utf-8'))
print(json.dumps(summary, ensure_ascii=False, indent=2))
```

## Происхождение профилей
Профили присоединяются только по явно указанному табельному номеру. Исходная типология поведения не переименовывается в четыре классических темперамента.


```python
profiles = pd.read_csv(out / 'profile_link_audit.csv')
print(profiles.groupby('profile_link_status').size().to_string())
print('Известных классических темпераментов:', profiles['temperament'].notna().sum())
```

## Содержательное сравнение
Категории различают совпадение, другой пример или деталь, непересказанный эпизод и неясность ASR. Лексическое сходство — отдельная поисковая подсказка, не вероятность лжи. Для каждой содержательной записи сохранены точные цитаты и время в content_comparison.json. Прослушивание спорных фрагментов требуется отдельно.


```python
review_path = out / 'content_comparison.csv'
review = pd.read_csv(review_path)
report = json.loads((out / 'content_comparison_report.json').read_text(encoding='utf-8'))
print('Историй со сверкой:', report['stories_reviewed'], '/', report['stories_total'])
print('Респондентов со сверкой:', report['participants_reviewed'])
print(review.groupby('relation').size().to_string())
print('Историй с кандидатами на противоречие:', report['stories_with_contradiction_candidates'])
print('Лексических кандидатов:', len(pd.read_csv(out / 'story_matches.csv')))
print('Несколько заметок по одному ответу не являются независимыми наблюдениями.')
```

## Проверка научных предпосылок
Без независимо размеченных условий отдельных ответов и валидированных темпераментов нельзя проверять различия лжи между темпераментами. Модуль содержит исполняемую LOSO-валидацию с предобработкой внутри каждого фолда и перестановочные тесты по индивидуальным контрастам. Тестовые синтетические данные не входят в научные результаты.


```python
for name in ('hypotheses.json', 'validation.json'):
    print(name)
    print(json.dumps(json.loads((out / name).read_text(encoding='utf-8')), ensure_ascii=False, indent=2))
```
